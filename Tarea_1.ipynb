## 3. Orbitas planetarias:

En el espacio, la orbita de un cuerpo alrededor de otro (como un planeta alrededor del Sol), no necesariamente es circular. En general, toma la forma de una elipse, con el cuerpo a veces mas cerca y otras mas lejos. Si tenemos la distancia $l_1$ de maxima aproximacion de un planeta al Sol (su perihelio), y su velocidad lineal $v_1$ en el perihelio, entonces cualquier otra propiedad de la orbita se puede calcular a partir de estas dos cantidades de la siguiente manera:

### b)

```python
#Primero Guardamos los valores de las orbitas menores 
d_T = float(input("Ingrese la distancia al Sol en metros"))
v_T = float(input("Ingrese la velocidad en el perihelio"))
```

```python
d_H = float(input("Ingrese la distancia al Sol"))
v_H = float(input("Ingrese la velocidad en el perihelio"))
```

```python
#definimos la funcion
def Orbita(a,b):
    G = 6.67430e-11      
    M = 1.989e30 
    c = (2 * G * M) / (a * b)
    raiz = (c**2) + 4 * (b**2 - ((2 * G * M) / a))
    
    v2 = (c - (np.sqrt(raiz)))/2
          
    l2 = (a*b)/v2
    d = 0.5*(a + l2)
    e = np.sqrt(a * l2)
    T = (2*np.pi*d*e)/(a*b)
    ex =(l2-a)/(l2+a)
    print(f"Distancia en el afelio: {l2:.3e} m")
    print(f"Velocidad en afelio: {v2:.3e} m/s")
    print(f"Periodo orbital: {T:.3e} s")
    print(f"Excentricidad orbital: {ex:.3f}")
```

## Problema 5

```python
import time # esta es para medir el tiempoo de ejecucion 

def madelung(L):
    start = time.time()
    # Creamos las coordenadas 
    r = np.arange(-L, L + 1)
    x, y, z = np.meshgrid(r, r, r, indexing='ij')

    #para no dividir entre cero
    m = (x != 0) | (y != 0) | (z != 0)

    R = np.sqrt(x**2 + y**2 + z**2)
    R = R[m]

    signos = np.power(-1.0, x + y + z)
    signos = signos[m]

    M = np.sum(signos / R)

    end = time.time()
    print(f"\nL = {L} => M = {M:.6f}")
    print(f"Tiempo de ejecución: {end - start:.3f} segundos")
    return M
```

```python
def triangulo_de_pascal(n): #Genera el triángulo de Pascal hasta la fila n.
    triangulo = []  

    for i in range(n):
        fila = [1] 
        if triangulo:  
            ultima_fila = triangulo[-1]  
            for j in range(len(ultima_fila) - 1):
                fila.append(ultima_fila[j] + ultima_fila[j + 1])  
            fila.append(1)  
        triangulo.append(fila)  
    return triangulo

# Pedir el número de filas
num_filas = int(input("Introduce el número de filas: "))
triangulo = triangulo_de_pascal(num_filas)
```
