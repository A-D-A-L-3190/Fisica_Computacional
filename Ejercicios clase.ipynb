```python
#generalizacion para cualquier matriz nxn
def Matrix(A,B):
    a=len(A[0])
    b=len(B)   
    R=np.zeros((a,b))
    if len(A[0]) == len(B):
        for i in range(len(A)):
            for j in range(len(B[0])):
                for k in range(len(B)):
                    R[i][j] += A[i][k] * B[k][j]
    else:
        print("Ramero, no se puedee!")
    return(R)
```

```python
#Primera aproximacion a la integral
#Regla de simpson

def f1(x):
    return (x**4) - 2*x +1
    
def Int1(f,a,b,N):
    h = (b-a)/N
    A=0
    for i in range(0,N):
        A += (f(a+(i*h)))*h
    return(A)
```
