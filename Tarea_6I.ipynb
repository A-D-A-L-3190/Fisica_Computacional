```python
V= 1.25

N=100
error = 1.0
exactitud = 1e-3

PHI_n = np.zeros((N+1, N+1))

PHI_n1 = np.copy(PHI_n)

PHI_n1[:,N] = V

Eps = 1 #8.854e-12

#Falta defini las cargas 
while error>exactitud:
    for i in range(N+1):
        for j in range (N+1):
            if i == 0 or i == N or j == N:
                PHI_n[i,j] = PHI_n1[i,j] 
            elif (0.2*N <= i <= 0.4*N) or (0.6*N <= i <= 0.8*N) or \
                 (0.2*N <= j <= 0.4*N) or (0.6*N <= j <= 0.8*N):
                PHI_n[i,j] =((-1)/(4*Eps))+(PHI_n1[i+1,j] + PHI_n1[i-1,j] + PHI_n1[i,j+1] + PHI_n1[i,j-1])/4
                
            else:
                PHI_n[i,j] =((-1)/(4*Eps))+(PHI_n1[i+1,j] + PHI_n1[i-1,j] + PHI_n1[i,j+1] + PHI_n1[i,j-1])/4

    error=np.max(np.abs(PHI_n1 - PHI_n))

    PHI_n1,PHI_n = PHI_n, PHI_n1

plt.figure(figsize=(6, 5))
plt.imshow(PHI_n1, origin='lower', cmap='viridis', extent=[0, N, 0, N])
plt.colorbar(label='Potencial eléctrico (V)')
plt.title('Distribución del potencial eléctrico')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
